```python

import os
import xmlschema
schema_source='sample_data'

#build schema repository
repo={}
for schema_file in os.scandir(schema_source):
    #schema=schema_file.name.split()[0] #+".xsd"
    #repo[schema]=xmlschema.XMLSchema(schema_file.name, base_url=schema_source)
    schema=xmlschema.XMLSchema(schema_file.name, base_url=schema_source)
    repo[schema.name]=schema
```
